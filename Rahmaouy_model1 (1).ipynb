```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

"""This script trains a Logistic Regression classifier using TF IDF word ngramfeatures """

# Loading the training and test data
train = pd.read_csv("data/train.csv")
test = pd.read_csv("data/test.csv")

# Separating the input text from the gold labels
train_text = train["TEXT"].astype(str)
train_label = train["LABEL"]

# Converting text into TF IDF features
# ngram_range=(1, 3)
vectorizer = TfidfVectorizer(lowercase=True,ngram_range=(1, 3),min_df=3,sublinear_tf=True)
F_train = vectorizer.fit_transform(train_text)

# Training the  Logistic regression classifier
model = LogisticRegression(max_iter=5000,class_weight="balanced",C=20.0,solver="liblinear")
model.fit(F_train, train_label)

# Transforming  the test text using the same vectorizer.
test_text = test["TEXT"].astype(str)
F_test = vectorizer.transform(test_text)

# Predicting labels.
predictions = model.predict(F_test)


submit = pd.DataFrame({"ID": test["ID"],"LABEL": predictions})
#submit.to_csv("/Users/fedouarahmaouy/Desktop/18submission.csv", index=False)
```
